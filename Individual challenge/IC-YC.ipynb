## Reorganization the data architectures and data cleaning

```python
import pandas as pd

file_path = "API_NY.GDP.PCAP.CD_DS2_en_csv_v2_85121.csv"


df_raw = pd.read_csv(file_path, header=None)

start_year = 1998
n_years = df_raw.shape[1] - 1
years = list(range(start_year, start_year + n_years))
column_names = ["Country"] + years
df_raw.columns = column_names


df_long = df_raw.melt(id_vars="Country", var_name="Year", value_name="GDP per capita")


df_long = df_long.dropna()
df_long["Year"] = df_long["Year"].astype(int)
df_long["GDP per capita"] = pd.to_numeric(df_long["GDP per capita"], errors="coerce")

df_long = df_long.sort_values(by=["Country", "Year"])


output_path = "GDP.csv"
df_long.to_excel(output_path, index=False)
print(f"Got it：{output_path}")
```
